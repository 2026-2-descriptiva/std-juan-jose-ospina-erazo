```python
import pandas as pd
import os


INPUT_FILE = "../data/raw.csv"
OUTPUT_FILE = "../submission/anonymized.csv"


def anonymize():

    df = pd.read_csv(INPUT_FILE)

    # eliminar datos personales directos
    columnas_privadas = [
        "name",
        "document_id",
        "email",
        "loyalty_card_number"
    ]

    df = df.drop(
        columns=[c for c in columnas_privadas if c in df.columns]
    )

    # crear ID anónimo
    df.insert(
        0,
        "anonymous_id",
        range(1, len(df)+1)
    )

    os.makedirs(
        "../submission",
        exist_ok=True
    )

    df.to_csv(
        OUTPUT_FILE,
        index=False
    )


anonymize()
```
